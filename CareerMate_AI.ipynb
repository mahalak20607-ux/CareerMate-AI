```python
# ============================================================
# CELL 3 - COMPLETE CAREERMATE AI APPLICATION
# ============================================================

import gradio as gr
import speech_recognition as sr
import tempfile
import os
import re
from pydub import AudioSegment


# ============================================================
# CAREER DATABASE
# ============================================================

CAREERS = {

    "Software Developer": {
        "keywords": [
            "python", "java", "javascript", "coding",
            "programming", "software", "developer",
            "web", "react", "node", "html", "css",
            "github", "git"
        ],

        "skills": [
            "Programming",
            "Data Structures",
            "Git & GitHub",
            "Web Development",
            "Problem Solving",
            "Database Basics"
        ],

        "roadmap": [
            "Learn Python or Java",
            "Learn Data Structures and Algorithms",
            "Learn HTML, CSS and JavaScript",
            "Learn Git and GitHub",
            "Learn React or another frontend framework",
            "Build 2–3 real-world projects",
            "Practice coding interview problems"
        ]
    },


    "Data Analyst": {
        "keywords": [
            "data", "excel", "statistics", "analysis",
            "analytics", "python", "sql", "power bi",
            "tableau", "charts", "numbers", "database"
        ],

        "skills": [
            "Excel",
            "SQL",
            "Python",
            "Statistics",
            "Data Visualization",
            "Power BI"
        ],

        "roadmap": [
            "Master Excel",
            "Learn SQL",
            "Learn basic statistics",
            "Learn Python for data analysis",
            "Practice Pandas and NumPy",
            "Learn Power BI or Tableau",
            "Create data-analysis projects"
        ]
    },


    "UI/UX Designer": {
        "keywords": [
            "design", "ui", "ux", "figma", "creative",
            "prototype", "wireframe", "visual",
            "user experience", "interface"
        ],

        "skills": [
            "Figma",
            "Wireframing",
            "Prototyping",
            "User Research",
            "Visual Design",
            "Design Thinking"
        ],

        "roadmap": [
            "Learn UI/UX fundamentals",
            "Learn Figma",
            "Practice wireframes",
            "Create interactive prototypes",
            "Study design systems",
            "Build a design portfolio",
            "Complete real-world UI/UX projects"
        ]
    },


    "Cybersecurity Analyst": {
        "keywords": [
            "security", "cyber", "hacking", "network",
            "linux", "firewall", "privacy", "threat",
            "ethical hacking", "penetration"
        ],

        "skills": [
            "Networking",
            "Linux",
            "Cybersecurity Fundamentals",
            "Threat Detection",
            "Security Tools",
            "Ethical Hacking"
        ],

        "roadmap": [
            "Learn computer networking",
            "Learn Linux",
            "Understand cybersecurity fundamentals",
            "Learn common security threats",
            "Practice with security labs",
            "Learn security monitoring",
            "Build cybersecurity projects"
        ]
    },


    "Cloud Engineer": {
        "keywords": [
            "cloud", "aws", "azure", "docker",
            "server", "deployment", "devops",
            "kubernetes", "linux", "cloud computing"
        ],

        "skills": [
            "Linux",
            "Cloud Computing",
            "AWS/Azure",
            "Docker",
            "Networking",
            "DevOps"
        ],

        "roadmap": [
            "Learn Linux basics",
            "Learn networking fundamentals",
            "Learn AWS or Azure",
            "Learn Docker",
            "Understand CI/CD",
            "Learn basic Kubernetes",
            "Deploy cloud projects"
        ]
    },


    "Digital Marketing Specialist": {
        "keywords": [
            "marketing", "social media", "instagram",
            "content", "seo", "advertising", "branding",
            "communication", "creative"
        ],

        "skills": [
            "SEO",
            "Content Creation",
            "Social Media",
            "Analytics",
            "Branding",
            "Communication"
        ],

        "roadmap": [
            "Learn digital marketing basics",
            "Learn SEO",
            "Practice social-media marketing",
            "Learn content creation",
            "Understand analytics",
            "Practice advertising campaigns",
            "Build a marketing portfolio"
        ]
    }
}


# ============================================================
# TEXT CLEANING
# ============================================================

def clean_text(text):

    if not text:
        return ""

    text = text.lower()

    text = re.sub(
        r"[^a-zA-Z0-9\s/&+-]",
        " ",
        text
    )

    text = re.sub(
        r"\s+",
        " ",
        text
    )

    return text.strip()


# ============================================================
# CAREER MATCHING
# ============================================================

def calculate_career_matches(text):

    cleaned = clean_text(text)

    results = []

    for career, data in CAREERS.items():

        matched_keywords = []

        for keyword in data["keywords"]:

            if keyword.lower() in cleaned:
                matched_keywords.append(keyword)

        score = len(matched_keywords)

        results.append({
            "career": career,
            "score": score,
            "matched": matched_keywords
        })

    results.sort(
        key=lambda item: item["score"],
        reverse=True
    )

    return results


# ============================================================
# SKILL EXTRACTION
# ============================================================

def extract_skills(text):

    cleaned = clean_text(text)

    skill_keywords = {

        "Python": ["python"],

        "Java": ["java"],

        "JavaScript": ["javascript"],

        "SQL": ["sql"],

        "Excel": ["excel"],

        "Power BI": ["power bi"],

        "Figma": ["figma"],

        "Git & GitHub": [
            "github",
            "git"
        ],

        "HTML/CSS": [
            "html",
            "css"
        ],

        "React": [
            "react"
        ],

        "Docker": [
            "docker"
        ],

        "AWS": [
            "aws"
        ],

        "Azure": [
            "azure"
        ],

        "Linux": [
            "linux"
        ],

        "Networking": [
            "network",
            "networking"
        ],

        "Cybersecurity": [
            "cybersecurity",
            "cyber",
            "security"
        ],

        "SEO": [
            "seo"
        ],

        "Communication": [
            "communication"
        ],

        "Data Analysis": [
            "data analysis",
            "analytics"
        ],

        "UI/UX": [
            "ui",
            "ux",
            "design"
        ]
    }

    found = []

    for skill, keywords in skill_keywords.items():

        for keyword in keywords:

            if keyword in cleaned:

                found.append(skill)

                break

    return list(dict.fromkeys(found))


# ============================================================
# CAREER READINESS
# ============================================================

def calculate_readiness(text, career_data):

    cleaned = clean_text(text)

    matched = 0

    for skill in career_data["skills"]:

        skill_words = skill.lower().split()

        if any(
            word in cleaned
            for word in skill_words
            if len(word) > 2
        ):

            matched += 1

    total = len(career_data["skills"])

    if total == 0:
        return 0

    score = int(
        (matched / total) * 100
    )

    return min(score, 100)


# ============================================================
# CAREER ANALYSIS
# ============================================================

def analyze_career(text):

    if not text or not text.strip():

        return """
        <div style="
            padding:25px;
            border-radius:18px;
            background:#fff7ed;
            border:2px solid #fed7aa;
            color:#9a3412;
        ">
            <h3>⚠️ Please enter some information</h3>

            <p>
            Tell CareerMate about your skills, interests,
            education or career goals.
            </p>
        </div>
        """

    matches = calculate_career_matches(text)

    best = matches[0]

    # If nothing matches
    if best["score"] == 0:

        return """
        <div style="
            padding:25px;
            border-radius:18px;
            background:#eff6ff;
            border:2px solid #bfdbfe;
            color:#1e3a8a;
        ">

            <h2>🔎 More Information Needed</h2>

            <p>
            CareerMate could not find enough matching
            career keywords.
            </p>

            <p>
            Try mentioning skills such as Python, SQL,
            Excel, Figma, cloud, cybersecurity, marketing,
            coding or design.
            </p>

        </div>
        """

    career = best["career"]

    career_data = CAREERS[career]

    readiness = calculate_readiness(
        text,
        career_data
    )

    current_skills = extract_skills(text)

    matched_keywords = best["matched"]

    # --------------------------------------------------------
    # MATCHED KEYWORDS
    # --------------------------------------------------------

    if matched_keywords:

        keyword_html = ""

        for keyword in matched_keywords:

            keyword_html += f"""
            <span style="
                display:inline-block;
                background:#e0e7ff;
                color:#3730a3;
                padding:7px 12px;
                margin:4px;
                border-radius:20px;
                font-size:12px;
                font-weight:600;
            ">
                {keyword}
            </span>
            """

    else:

        keyword_html = """
        <span style="color:#64748b;">
            General interest detected
        </span>
        """


    # --------------------------------------------------------
    # DETECTED SKILLS
    # --------------------------------------------------------

    if current_skills:

        skills_html = ""

        for skill in current_skills:

            skills_html += f"""
            <span style="
                display:inline-block;
                background:#dcfce7;
                color:#166534;
                padding:7px 12px;
                margin:4px;
                border-radius:20px;
                font-size:12px;
                font-weight:600;
            ">
                ✓ {skill}
            </span>
            """

    else:

        skills_html = """
        <span style="color:#64748b;">
            No specific technical skills detected yet.
        </span>
        """


    # --------------------------------------------------------
    # SKILL GAP
    # --------------------------------------------------------

    missing_skills = []

    for skill in career_data["skills"]:

        if skill not in current_skills:

            missing_skills.append(skill)


    if missing_skills:

        gap_html = ""

        for skill in missing_skills:

            gap_html += f"""
            <div style="
                padding:8px 0;
                color:#475569;
            ">
                <span style="color:#f59e0b;">
                    ●
                </span>
                {skill}
            </div>
            """

    else:

        gap_html = """
        <div style="
            color:#15803d;
            font-weight:700;
        ">
            🎉 Excellent skill coverage!
        </div>
        """


    # --------------------------------------------------------
    # ROADMAP
    # --------------------------------------------------------

    roadmap_html = ""

    for index, step in enumerate(
        career_data["roadmap"],
        start=1
    ):

        roadmap_html += f"""
        <div style="
            display:flex;
            align-items:flex-start;
            gap:12px;
            margin:12px 0;
        ">

            <div style="
                min-width:30px;
                height:30px;
                border-radius:50%;
                background:linear-gradient(
                    135deg,
                    #4f46e5,
                    #7c3aed
                );
                color:white;
                display:flex;
                align-items:center;
                justify-content:center;
                font-weight:700;
                font-size:12px;
            ">
                {index}
            </div>

            <div style="
                padding-top:5px;
                color:#334155;
            ">
                {step}
            </div>

        </div>
        """


    # --------------------------------------------------------
    # READINESS LEVEL
    # --------------------------------------------------------

    if readiness >= 75:

        level = "🚀 Strong Career Readiness"
        level_color = "#15803d"
        level_bg = "#dcfce7"

    elif readiness >= 50:

        level = "⭐ Good Starting Point"
        level_color = "#b45309"
        level_bg = "#fef3c7"

    elif readiness >= 25:

        level = "📈 Beginner – Keep Learning"
        level_color = "#0369a1"
        level_bg = "#e0f2fe"

    else:

        level = "🌱 Beginner – Build Your Foundation"
        level_color = "#7c3aed"
        level_bg = "#ede9fe"


    # ========================================================
    # FINAL RESULT
    # ========================================================

    result = f"""

    <div style="
        font-family:Arial,sans-serif;
        color:#334155;
    ">


        <!-- RECOMMENDED CAREER -->

        <div style="
            background:linear-gradient(
                135deg,
                #4f46e5,
                #7c3aed,
                #2563eb
            );
            color:white;
            padding:28px;
            border-radius:22px;
            margin-bottom:18px;
            box-shadow:
                0 12px 30px rgba(79,70,229,0.20);
        ">

            <div style="
                font-size:13px;
                opacity:0.85;
                font-weight:700;
                letter-spacing:1px;
            ">
                RECOMMENDED CAREER
            </div>

            <h1 style="
                margin:8px 0;
                color:white !important;
                font-size:30px !important;
            ">
                💼 {career}
            </h1>

            <p style="
                margin:0;
                color:white !important;
                opacity:0.9;
            ">
                Career recommendation based on your
                skills and interests.
            </p>

        </div>


        <!-- READINESS -->

        <div style="
            background:white;
            border:1px solid #e2e8f0;
            border-radius:20px;
            padding:22px;
            margin-bottom:18px;
            box-shadow:0 6px 20px rgba(0,0,0,0.05);
        ">

            <div style="
                display:flex;
                justify-content:space-between;
                align-items:center;
            ">

                <div>

                    <div style="
                        color:#64748b;
                        font-size:13px;
                        font-weight:700;
                    ">
                        CAREER READINESS
                    </div>

                    <div style="
                        font-size:34px;
                        font-weight:800;
                        color:#312e81;
                        margin-top:5px;
                    ">
                        {readiness}%
                    </div>

                </div>

                <div style="
                    background:{level_bg};
                    color:{level_color};
                    padding:10px 15px;
                    border-radius:20px;
                    font-weight:700;
                    font-size:13px;
                ">
                    {level}
                </div>

            </div>

            <div style="
                margin-top:15px;
                height:10px;
                background:#e2e8f0;
                border-radius:10px;
                overflow:hidden;
            ">

                <div style="
                    width:{readiness}%;
                    height:100%;
                    background:linear-gradient(
                        90deg,
                        #4f46e5,
                        #7c3aed
                    );
                    border-radius:10px;
                "></div>

            </div>

        </div>


        <!-- MATCHING -->

        <div style="
            background:white;
            border:1px solid #e2e8f0;
            border-radius:20px;
            padding:22px;
            margin-bottom:18px;
        ">

            <h3 style="
                color:#312e81 !important;
                margin-top:0;
            ">
                🔎 Matching Interests
            </h3>

            <div>
                {keyword_html}
            </div>

        </div>


        <!-- DETECTED SKILLS -->

        <div style="
            background:white;
            border:1px solid #e2e8f0;
            border-radius:20px;
            padding:22px;
            margin-bottom:18px;
        ">

            <h3 style="
                color:#166534 !important;
                margin-top:0;
            ">
                🧠 Skills Detected
            </h3>

            <div>
                {skills_html}
            </div>

        </div>


        <!-- SKILL GAP -->

        <div style="
            background:#fffbeb;
            border:1px solid #fde68a;
            border-radius:20px;
            padding:22px;
            margin-bottom:18px;
        ">

            <h3 style="
                color:#92400e !important;
                margin-top:0;
            ">
                📊 Recommended Skill Improvements
            </h3>

            {gap_html}

        </div>


        <!-- ROADMAP -->

        <div style="
            background:white;
            border:1px solid #e2e8f0;
            border-radius:20px;
            padding:22px;
            margin-bottom:18px;
        ">

            <h3 style="
                color:#312e81 !important;
                margin-top:0;
            ">
                🗺️ Personalized Learning Roadmap
            </h3>

            {roadmap_html}

        </div>


        <!-- ADVICE -->

        <div style="
            background:linear-gradient(
                135deg,
                #eef2ff,
                #f5f3ff
            );
            border:1px solid #c7d2fe;
            border-radius:20px;
            padding:22px;
        ">

            <h3 style="
                color:#312e81 !important;
                margin-top:0;
            ">
                💡 Career Advice
            </h3>

            <p style="
                color:#475569;
                line-height:1.7;
            ">
                Focus on the skills listed in your skill gap,
                create practical projects, and gradually build
                a professional portfolio. Projects and practical
                experience can strengthen your career profile.
            </p>

        </div>

    </div>
    """

    return result


# ============================================================
# AUDIO → WAV
# ============================================================

def convert_to_wav(audio_file):

    if audio_file is None:

        return None

    try:

        audio = AudioSegment.from_file(
            audio_file
        )

        wav_path = tempfile.NamedTemporaryFile(
            delete=False,
            suffix=".wav"
        ).name

        audio.export(
            wav_path,
            format="wav"
        )

        return wav_path

    except Exception as e:

        print(
            "Audio conversion error:",
            e
        )

        return None


# ============================================================
# SPEECH TO TEXT
# ============================================================

def speech_to_text(audio_file):

    if audio_file is None:

        return ""

    wav_path = convert_to_wav(
        audio_file
    )

    if wav_path is None:

        return ""

    recognizer = sr.Recognizer()

    try:

        with sr.AudioFile(
            wav_path
        ) as source:

            audio_data = recognizer.record(
                source
            )

        text = recognizer.recognize_google(
            audio_data
        )

        return text

    except sr.UnknownValueError:

        return (
            "Unable to understand the voice input."
        )

    except sr.RequestError:

        return (
            "Speech recognition service is unavailable."
        )

    except Exception as e:

        return (
            f"Voice processing error: {str(e)}"
        )

    finally:

        try:

            if os.path.exists(wav_path):

                os.remove(wav_path)

        except:

            pass


# ============================================================
# VOICE ANALYSIS
# ============================================================

def analyze_voice(audio):

    if audio is None:

        return (
            """
            <div style="
                padding:20px;
                border-radius:16px;
                background:#fff7ed;
                border:2px solid #fed7aa;
                color:#9a3412;
            ">
                ⚠️ Please record or upload your voice.
            </div>
            """,
            ""
        )

    transcript = speech_to_text(
        audio
    )

    if not transcript:

        return (
            """
            <div style="
                padding:20px;
                border-radius:16px;
                background:#fff7ed;
                color:#9a3412;
            ">
                ⚠️ Could not process the voice input.
            </div>
            """,
            ""
        )

    if transcript.startswith(
        "Unable to understand"
    ):

        return (
            f"""
            <div style="
                padding:20px;
                border-radius:16px;
                background:#fff7ed;
                color:#9a3412;
            ">
                ⚠️ {transcript}
            </div>
            """,
            transcript
        )

    if transcript.startswith(
        "Speech recognition"
    ):

        return (
            f"""
            <div style="
                padding:20px;
                border-radius:16px;
                background:#fff7ed;
                color:#9a3412;
            ">
                ⚠️ {transcript}
            </div>
            """,
            transcript
        )

    analysis = analyze_career(
        transcript
    )

    return (
        analysis,
        transcript
    )


# ============================================================
# EXAMPLE TEXT
# ============================================================

def example_text():

    return """
I am a computer science student. I know basic Python,
HTML, CSS and GitHub. I enjoy coding and creating websites.
I am interested in software development and technology.
I also want to improve my programming skills.
"""


# ============================================================
# PROFESSIONAL GRADIO CSS
# ============================================================

css = """

/* MAIN PAGE */

.gradio-container {
    max-width: 1200px !important;
    margin: auto !important;
}


/* BODY */

body {
    background:
        linear-gradient(
            135deg,
            #eef2ff 0%,
            #f8f7ff 45%,
            #eff6ff 100%
        ) !important;
}


/* TABS */

button[role="tab"] {
    font-weight: 700 !important;
}


/* INPUTS */

textarea,
input {
    border-radius: 14px !important;
}


/* BUTTONS */

button {
    border-radius: 12px !important;
    font-weight: 700 !important;
}


/* MARKDOWN */

.markdown-text {
    color: #334155 !important;
}


/* FOOTER */

.footer-text {
    text-align: center;
}

"""


# ============================================================
# BUILD GRADIO APPLICATION
# ============================================================

with gr.Blocks(
    css=css,
    title="CareerMate AI"
) as app:


    # ========================================================
    # HERO
    # ========================================================

    gr.HTML("""
    <div style="
        padding:35px;
        margin-bottom:25px;
        border-radius:25px;

        background:
            linear-gradient(
                135deg,
                #4f46e5,
                #7c3aed,
                #2563eb
            );

        color:white;

        box-shadow:
            0 15px 40px rgba(79,70,229,0.25);
    ">

        <div style="
            font-size:42px;
            font-weight:800;
            color:white;
            margin-bottom:8px;
        ">
            🎯 CareerMate AI
        </div>

        <div style="
            font-size:20px;
            font-weight:600;
            color:white;
            margin-bottom:10px;
        ">
            Intelligent Career & Skill Recommendation System
        </div>

        <div style="
            font-size:15px;
            color:white;
            opacity:0.9;
            line-height:1.6;
        ">
            Discover suitable career paths using your
            skills, interests, education and goals.
        </div>

    </div>
    """)


    # ========================================================
    # TEXT CAREER ANALYSIS
    # ========================================================

    with gr.Tab("📝 Text Career Analysis"):

        gr.Markdown("""
        ## 📝 Tell CareerMate About Yourself

        Enter your **skills, interests, education,
        experience or career goals**.
        """)

        with gr.Row():

            with gr.Column(
                scale=1
            ):

                text_input = gr.Textbox(
                    label="About You",
                    placeholder=(
                        "Example: I know Python, HTML and CSS. "
                        "I enjoy coding, websites and technology..."
                    ),
                    lines=9
                )

                with gr.Row():

                    analyze_btn = gr.Button(
                        "🔍 Analyze My Career",
                        variant="primary"
                    )

                    example_btn = gr.Button(
                        "✨ Load Example"
                    )


            with gr.Column(
                scale=1
            ):

                text_output = gr.HTML(
                    value="""
                    <div style="
                        padding:25px;
                        text-align:center;
                        background:white;
                        border-radius:20px;
                        border:1px solid #e2e8f0;
                    ">

                        <div style="
                            font-size:45px;
                        ">
                            🎯
                        </div>

                        <h2 style="
                            color:#312e81;
                        ">
                            Your Career Analysis
                        </h2>

                        <p style="
                            color:#64748b;
                        ">
                            Enter your details and click
                            Analyze My Career.
                        </p>

                    </div>
                    """
                )


        analyze_btn.click(
            fn=analyze_career,
            inputs=text_input,
            outputs=text_output
        )

        example_btn.click(
            fn=example_text,
            inputs=None,
            outputs=text_input
        )


    # ========================================================
    # VOICE CAREER ANALYSIS
    # ========================================================

    with gr.Tab("🎙️ Voice Career Analysis"):

        gr.Markdown("""
        ## 🎙️ Speak About Your Career Interests

        Record your voice or upload an audio file.
        CareerMate will analyze the spoken information
        and recommend a suitable career path.
        """)

        with gr.Row():

            with gr.Column(
                scale=1
            ):

                voice_input = gr.Audio(
                    sources=[
                        "microphone",
                        "upload"
                    ],
                    type="filepath",
                    label="🎙️ Record or Upload Voice"
                )

                voice_btn = gr.Button(
                    "🎯 Analyze My Career",
                    variant="primary"
                )


            with gr.Column(
                scale=1
            ):

                transcript_output = gr.Textbox(
                    label="📝 Voice Transcript",
                    lines=5,
                    interactive=False
                )

                voice_output = gr.HTML(
                    value="""
                    <div style="
                        padding:25px;
                        text-align:center;
                        background:white;
                        border-radius:20px;
                        border:1px solid #e2e8f0;
                    ">

                        <div style="
                            font-size:45px;
                        ">
                            🎙️
                        </div>

                        <h2 style="
                            color:#312e81;
                        ">
                            Voice Career Analysis
                        </h2>

                        <p style="
                            color:#64748b;
                        ">
                            Record your voice to begin.
                        </p>

                    </div>
                    """
                )


        voice_btn.click(
            fn=analyze_voice,
            inputs=voice_input,
            outputs=[
                voice_output,
                transcript_output
            ]
        )


    # ========================================================
    # CAREER DIRECTORY
    # ========================================================

    with gr.Tab("💼 Career Directory"):


        gr.HTML("""
        <div style="
            padding:10px 5px 25px 5px;
            text-align:center;
        ">

            <h1 style="
                color:#312e81 !important;
                font-size:32px !important;
                font-weight:800 !important;
                margin-bottom:8px;
            ">
                💼 Explore Career Paths
            </h1>

            <p style="
                color:#64748b !important;
                font-size:16px !important;
                margin:0;
            ">
                Explore modern career paths and the skills
                required to build your professional future.
            </p>

        </div>
        """)


        gr.HTML("""
        <div style="
            display:grid;
            grid-template-columns:
                repeat(auto-fit,minmax(280px,1fr));
            gap:22px;
            padding:10px 0 25px 0;
        ">


        <!-- SOFTWARE DEVELOPER -->

        <div style="
            background:linear-gradient(
                135deg,#eef2ff,#ffffff
            );
            border:2px solid #c7d2fe;
            border-radius:22px;
            padding:25px;
            box-shadow:
                0 10px 25px rgba(79,70,229,0.12);
        ">

            <div style="
                width:58px;
                height:58px;
                background:linear-gradient(
                    135deg,#4f46e5,#7c3aed
                );
                border-radius:16px;
                display:flex;
                align-items:center;
                justify-content:center;
                font-size:28px;
                margin-bottom:18px;
            ">
                💻
            </div>

            <h2 style="
                color:#312e81 !important;
                font-size:21px !important;
                font-weight:800 !important;
                margin:0 0 12px 0;
            ">
                Software Developer
            </h2>

            <p style="
                color:#64748b !important;
                font-size:12px !important;
                font-weight:700 !important;
            ">
                CORE SKILLS
            </p>

            <div style="
                line-height:2.5;
            ">

                <span style="
                    background:#e0e7ff;
                    color:#3730a3;
                    padding:7px 11px;
                    border-radius:20px;
                    margin:3px;
                    font-size:12px;
                    font-weight:600;
                ">Programming</span>

                <span style="
                    background:#ede9fe;
                    color:#6d28d9;
                    padding:7px 11px;
                    border-radius:20px;
                    margin:3px;
                    font-size:12px;
                    font-weight:600;
                ">Data Structures</span>

                <span style="
                    background:#dbeafe;
                    color:#1d4ed8;
                    padding:7px 11px;
                    border-radius:20px;
                    margin:3px;
                    font-size:12px;
                    font-weight:600;
                ">Git & GitHub</span>

                <span style="
                    background:#e0f2fe;
                    color:#0369a1;
                    padding:7px 11px;
                    border-radius:20px;
                    margin:3px;
                    font-size:12px;
                    font-weight:600;
                ">Web Development</span>

                <span style="
                    background:#f3e8ff;
                    color:#7e22ce;
                    padding:7px 11px;
                    border-radius:20px;
                    margin:3px;
                    font-size:12px;
                    font-weight:600;
                ">Problem Solving</span>

                <span style="
                    background:#e0e7ff;
                    color:#3730a3;
                    padding:7px 11px;
                    border-radius:20px;
                    margin:3px;
                    font-size:12px;
                    font-weight:600;
                ">Database Basics</span>

            </div>

        </div>


        <!-- DATA ANALYST -->

        <div style="
            background:linear-gradient(
                135deg,#ecfeff,#ffffff
            );
            border:2px solid #a5f3fc;
            border-radius:22px;
            padding:25px;
            box-shadow:
                0 10px 25px rgba(6,182,212,0.12);
        ">

            <div style="
                width:58px;
                height:58px;
                background:linear-gradient(
                    135deg,#0891b2,#2563eb
                );
                border-radius:16px;
                display:flex;
                align-items:center;
                justify-content:center;
                font-size:28px;
                margin-bottom:18px;
            ">
                📊
            </div>

            <h2 style="
                color:#0f4c5c !important;
                font-size:21px !important;
                font-weight:800 !important;
                margin:0 0 12px 0;
            ">
                Data Analyst
            </h2>

            <p style="
                color:#64748b !important;
                font-size:12px !important;
                font-weight:700 !important;
            ">
                CORE SKILLS
            </p>

            <div style="
                line-height:2.5;
            ">

                <span style="
                    background:#cffafe;
                    color:#0e7490;
                    padding:7px 11px;
                    border-radius:20px;
                    margin:3px;
                    font-size:12px;
                    font-weight:600;
                ">Excel</span>

                <span style="
                    background:#dbeafe;
                    color:#1d4ed8;
                    padding:7px 11px;
                    border-radius:20px;
                    margin:3px;
                    font-size:12px;
                    font-weight:600;
                ">SQL</span>

                <span style="
                    background:#dcfce7;
                    color:#15803d;
                    padding:7px 11px;
                    border-radius:20px;
                    margin:3px;
                    font-size:12px;
                    font-weight:600;
                ">Python</span>

                <span style="
                    background:#fef3c7;
                    color:#b45309;
                    padding:7px 11px;
                    border-radius:20px;
                    margin:3px;
                    font-size:12px;
                    font-weight:600;
                ">Statistics</span>

                <span style="
                    background:#ede9fe;
                    color:#6d28d9;
                    padding:7px 11px;
                    border-radius:20px;
                    margin:3px;
                    font-size:12px;
                    font-weight:600;
                ">Data Visualization</span>

                <span style="
                    background:#e0e7ff;
                    color:#3730a3;
                    padding:7px 11px;
                    border-radius:20px;
                    margin:3px;
                    font-size:12px;
                    font-weight:600;
                ">Power BI</span>

            </div>

        </div>


        <!-- UI UX -->

        <div style="
            background:linear-gradient(
                135deg,#fdf2f8,#ffffff
            );
            border:2px solid #f9a8d4;
            border-radius:22px;
            padding:25px;
            box-shadow:
                0 10px 25px rgba(236,72,153,0.12);
        ">

            <div style="
                width:58px;
                height:58px;
                background:linear-gradient(
                    135deg,#db2777,#9333ea
                );
                border-radius:16px;
                display:flex;
                align-items:center;
                justify-content:center;
                font-size:28px;
                margin-bottom:18px;
            ">
                🎨
            </div>

            <h2 style="
                color:#831843 !important;
                font-size:21px !important;
                font-weight:800 !important;
                margin:0 0 12px 0;
            ">
                UI/UX Designer
            </h2>

            <p style="
                color:#64748b !important;
                font-size:12px !important;
                font-weight:700 !important;
            ">
                CORE SKILLS
            </p>

            <div style="
                line-height:2.5;
            ">

                <span style="
                    background:#fce7f3;
                    color:#be185d;
                    padding:7px 11px;
                    border-radius:20px;
                    margin:3px;
                    font-size:12px;
                    font-weight:600;
                ">Figma</span>

                <span style="
                    background:#f3e8ff;
                    color:#7e22ce;
                    padding:7px 11px;
                    border-radius:20px;
                    margin:3px;
                    font-size:12px;
                    font-weight:600;
                ">Wireframing</span>

                <span style="
                    background:#ede9fe;
                    color:#6d28d9;
                    padding:7px 11px;
                    border-radius:20px;
                    margin:3px;
                    font-size:12px;
                    font-weight:600;
                ">Prototyping</span>

                <span style="
                    background:#fce7f3;
                    color:#9d174d;
                    padding:7px 11px;
                    border-radius:20px;
                    margin:3px;
                    font-size:12px;
                    font-weight:600;
                ">User Research</span>

                <span style="
                    background:#fae8ff;
                    color:#a21caf;
                    padding:7px 11px;
                    border-radius:20px;
                    margin:3px;
                    font-size:12px;
                    font-weight:600;
                ">Visual Design</span>

                <span style="
                    background:#fef3c7;
                    color:#b45309;
                    padding:7px 11px;
                    border-radius:20px;
                    margin:3px;
                    font-size:12px;
                    font-weight:600;
                ">Design Thinking</span>

            </div>

        </div>


        <!-- CYBERSECURITY -->

        <div style="
            background:linear-gradient(
                135deg,#ecfdf5,#ffffff
            );
            border:2px solid #86efac;
            border-radius:22px;
            padding:25px;
            box-shadow:
                0 10px 25px rgba(16,185,129,0.12);
        ">

            <div style="
                width:58px;
                height:58px;
                background:linear-gradient(
                    135deg,#059669,#047857
                );
                border-radius:16px;
                display:flex;
                align-items:center;
                justify-content:center;
                font-size:28px;
                margin-bottom:18px;
            ">
                🛡️
            </div>

            <h2 style="
                color:#065f46 !important;
                font-size:21px !important;
                font-weight:800 !important;
                margin:0 0 12px 0;
            ">
                Cybersecurity Analyst
            </h2>

            <p style="
                color:#64748b !important;
                font-size:12px !important;
                font-weight:700 !important;
            ">
                CORE SKILLS
            </p>

            <div style="
                line-height:2.5;
            ">

                <span style="
                    background:#d1fae5;
                    color:#047857;
                    padding:7px 11px;
                    border-radius:20px;
                    margin:3px;
                    font-size:12px;
                    font-weight:600;
                ">Networking</span>

                <span style="
                    background:#dcfce7;
                    color:#166534;
                    padding:7px 11px;
                    border-radius:20px;
                    margin:3px;
                    font-size:12px;
                    font-weight:600;
                ">Linux</span>

                <span style="
                    background:#d1fae5;
                    color:#065f46;
                    padding:7px 11px;
                    border-radius:20px;
                    margin:3px;
                    font-size:12px;
                    font-weight:600;
                ">Cybersecurity</span>

                <span style="
                    background:#fef3c7;
                    color:#92400e;
                    padding:7px 11px;
                    border-radius:20px;
                    margin:3px;
                    font-size:12px;
                    font-weight:600;
                ">Threat Detection</span>

                <span style="
                    background:#ccfbf1;
                    color:#115e59;
                    padding:7px 11px;
                    border-radius:20px;
                    margin:3px;
                    font-size:12px;
                    font-weight:600;
                ">Security Tools</span>

                <span style="
                    background:#dbeafe;
                    color:#1e40af;
                    padding:7px 11px;
                    border-radius:20px;
                    margin:3px;
                    font-size:12px;
                    font-weight:600;
                ">Ethical Hacking</span>

            </div>

        </div>


        <!-- CLOUD ENGINEER -->

        <div style="
            background:linear-gradient(
                135deg,#eff6ff,#ffffff
            );
            border:2px solid #93c5fd;
            border-radius:22px;
            padding:25px;
            box-shadow:
                0 10px 25px rgba(37,99,235,0.12);
        ">

            <div style="
                width:58px;
                height:58px;
                background:linear-gradient(
                    135deg,#2563eb,#06b6d4
                );
                border-radius:16px;
                display:flex;
                align-items:center;
                justify-content:center;
                font-size:28px;
                margin-bottom:18px;
            ">
                ☁️
            </div>

            <h2 style="
                color:#1e3a8a !important;
                font-size:21px !important;
                font-weight:800 !important;
                margin:0 0 12px 0;
            ">
                Cloud Engineer
            </h2>

            <p style="
                color:#64748b !important;
                font-size:12px !important;
                font-weight:700 !important;
            ">
                CORE SKILLS
            </p>

            <div style="
                line-height:2.5;
            ">

                <span style="
                    background:#dbeafe;
                    color:#1d4ed8;
                    padding:7px 11px;
                    border-radius:20px;
                    margin:3px;
                    font-size:12px;
                    font-weight:600;
                ">Linux</span>

                <span style="
                    background:#e0f2fe;
                    color:#0369a1;
                    padding:7px 11px;
                    border-radius:20px;
                    margin:3px;
                    font-size:12px;
                    font-weight:600;
                ">Cloud Computing</span>

                <span style="
                    background:#dbeafe;
                    color:#1e40af;
                    padding:7px 11px;
                    border-radius:20px;
                    margin:3px;
                    font-size:12px;
                    font-weight:600;
                ">AWS</span>

                <span style="
                    background:#cffafe;
                    color:#0e7490;
                    padding:7px 11px;
                    border-radius:20px;
                    margin:3px;
                    font-size:12px;
                    font-weight:600;
                ">Azure</span>

                <span style="
                    background:#ede9fe;
                    color:#6d28d9;
                    padding:7px 11px;
                    border-radius:20px;
                    margin:3px;
                    font-size:12px;
                    font-weight:600;
                ">Docker</span>

                <span style="
                    background:#e0e7ff;
                    color:#3730a3;
                    padding:7px 11px;
                    border-radius:20px;
                    margin:3px;
                    font-size:12px;
                    font-weight:600;
                ">DevOps</span>

            </div>

        </div>


        <!-- DIGITAL MARKETING -->

        <div style="
            background:linear-gradient(
                135deg,#fff7ed,#ffffff
            );
            border:2px solid #fdba74;
            border-radius:22px;
            padding:25px;
            box-shadow:
                0 10px 25px rgba(249,115,22,0.12);
        ">

            <div style="
                width:58px;
                height:58px;
                background:linear-gradient(
                    135deg,#f97316,#ea580c
                );
                border-radius:16px;
                display:flex;
                align-items:center;
                justify-content:center;
                font-size:28px;
                margin-bottom:18px;
            ">
                📱
            </div>

            <h2 style="
                color:#9a3412 !important;
                font-size:21px !important;
                font-weight:800 !important;
                margin:0 0 12px 0;
            ">
                Digital Marketing Specialist
            </h2>

            <p style="
                color:#64748b !important;
                font-size:12px !important;
                font-weight:700 !important;
            ">
                CORE SKILLS
            </p>

            <div style="
                line-height:2.5;
            ">

                <span style="
                    background:#ffedd5;
                    color:#c2410c;
                    padding:7px 11px;
                    border-radius:20px;
                    margin:3px;
                    font-size:12px;
                    font-weight:600;
                ">SEO</span>

                <span style="
                    background:#fef3c7;
                    color:#b45309;
                    padding:7px 11px;
                    border-radius:20px;
                    margin:3px;
                    font-size:12px;
                    font-weight:600;
                ">Content Creation</span>

                <span style="
                    background:#fce7f3;
                    color:#be185d;
                    padding:7px 11px;
                    border-radius:20px;
                    margin:3px;
                    font-size:12px;
                    font-weight:600;
                ">Social Media</span>

                <span style="
                    background:#ffedd5;
                    color:#c2410c;
                    padding:7px 11px;
                    border-radius:20px;
                    margin:3px;
                    font-size:12px;
                    font-weight:600;
                ">Analytics</span>

                <span style="
                    background:#f3e8ff;
                    color:#7e22ce;
                    padding:7px 11px;
                    border-radius:20px;
                    margin:3px;
                    font-size:12px;
                    font-weight:600;
                ">Branding</span>

                <span style="
                    background:#dbeafe;
                    color:#1d4ed8;
                    padding:7px 11px;
                    border-radius:20px;
                    margin:3px;
                    font-size:12px;
                    font-weight:600;
                ">Communication</span>

            </div>

        </div>

        </div>
        """)


        # ====================================================
        # CAREER DIRECTORY FOOTER
        # ====================================================

        gr.HTML("""
        <div style="
            text-align:center;
            margin-top:10px;
            padding:25px;
            border-radius:20px;

            background:linear-gradient(
                135deg,
                #eef2ff,
                #f5f3ff,
                #eff6ff
            );

            border:2px solid #c7d2fe;
        ">

            <div style="
                font-size:23px;
                font-weight:800;
                color:#312e81;
            ">
                🎯 Find the Career That Matches You
            </div>

            <div style="
                margin-top:8px;
                color:#475569;
                font-size:14px;
            ">
                Use Text Career Analysis or Voice Career
                Analysis to discover your personalized career path.
            </div>

        </div>
        """)


    # ========================================================
    # FOOTER
    # ========================================================

    gr.HTML("""
    <div style="
        text-align:center;
        padding:25px 10px 10px 10px;
        color:#64748b;
        font-size:13px;
    ">

        <b style="color:#4f46e5;">
            🎯 CareerMate AI
        </b>

        <br>

        Intelligent Career & Skill Recommendation System

        <br><br>

        🎓 Text + Voice + NLP

    </div>
    """)


# ============================================================
# LAUNCH APPLICATION
# ============================================================

print("")
print("==============================================")
print("       🎯 CareerMate AI")
print("==============================================")
print("🚀 Starting application...")
print("🌐 A public Gradio link will appear below.")
print("==============================================")
print("")


app.launch(
    share=True,
    debug=True
)
```
